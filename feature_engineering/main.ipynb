# For Sales Agent

1. Month ---> order_date
2. Quarter --> order_Date
3. day_of_week --> order_Date
4. year --> order_date

```python
master_df["order_date"] = pd.to_datetime(master_df["order_date"])
master_df['order_year'] = master_df['order_date'].dt.year
master_df['order_month'] = master_df['order_date'].dt.month_name()
master_df['order_quarter'] = master_df['order_date'].dt.quarter
master_df['order_day_of_week'] = master_df['order_date'].dt.day_name()
```

```python
master_df.head()
```

```python
monthly_revenue_df = (master_df.groupby(['order_year','order_month'],as_index=False).agg(
    monthly_revenue = ('payment_value','sum')
))

monthly_orders_df = (master_df.groupby(['order_month','order_year'],as_index=False).agg(
    monthly_orders = ('order_id','count')
))

sales_df = monthly_revenue_df.merge(
    monthly_orders_df, on = ['order_year','order_month']
)
```

```python
sales_df
```

```python
sales_df["holiday_flag"] = sales_df["order_month"].isin(['January', 'April', 'August', 'October', 'November', 'December']).astype(int)
```

```python
sales_df["festival_flag"] = sales_df["order_month"].isin(['August', 'October', 'November']).astype(int)
```

```python
import numpy as np

sales_df["discount_percentage"] = 0

# Festival Months (High Discount)

festival_mask = sales_df["festival_flag"] == 1

sales_df.loc[festival_mask, "discount_percentage"] = np.random.choice(
    [10, 15, 20, 25, 30],
    size=festival_mask.sum(),
    p=[0.15, 0.25, 0.30, 0.20, 0.10]
)

# Holiday Months (Non-Festival means less discount)
holiday_mask = (
    (sales_df["holiday_flag"] == 1) &
    (sales_df["festival_flag"] == 0)
)

sales_df.loc[holiday_mask, "discount_percentage"] = np.random.choice(
    [5, 10, 15],
    size=holiday_mask.sum(),
    p=[0.50, 0.35, 0.15]
)

# Normal Months (No discount)
normal_mask = (
    (sales_df["holiday_flag"] == 0) &
    (sales_df["festival_flag"] == 0)
)

sales_df.loc[normal_mask, "discount_percentage"] = np.random.choice(
    [0, 5, 10],
    size=normal_mask.sum(),
    p=[0.60, 0.30, 0.10]
)

```

# Customer Agent
Grouping with the customer_unique_id
1. total_orders --> order_id 
2. total_spend  --> payment_value
3. avg_order_value --> payment_value (mean)
4. recency_days -->order_date
5. purchase_frequency --> total_order / active months
6. avg_days_between_orders  --> order_date

```python
master_df = master_df.sort_values(by=['customer_unique_id','order_date'])
```

```python
refrence_date = master_df['order_date'].max()
```

```python
customer_features = (master_df.groupby('customer_unique_id',as_index=False).agg(
    customer_state = ('customer_state','first'),
    customer_city = ('customer_city','first'),
    total_orders = ('order_id','count'),
    total_spend = ('payment_value','sum'),
    avg_order_value = ('payment_value','mean'),
    first_purchase = ('order_date','min'),
    last_purchase = ('order_date','max'),
    avg_review_score = ('review_score','mean')

))
```

```python
customer_features['recency_days'] = (refrence_date - customer_features['last_purchase']).dt.days
```

```python
customer_features['customer_active_days'] = (customer_features['last_purchase'] - customer_features['first_purchase']).dt.days + 1
```

```python
customer_features['purchase_frequency'] =( customer_features['total_orders']/(customer_features['customer_active_days']/30))
```

```python
import numpy as np

# Replace any infinite values

customer_features['purchase_frequency'] = (customer_features['purchase_frequency'].replace([np.inf , -np.inf],0).fillna(0))
```

```python
master_df.info()
```

```python
customer_features.info()
```

```python
avg_gap = (
    master_df
    .sort_values(["customer_unique_id", "order_date"])
)

avg_gap["gap"] = (
    avg_gap
    .groupby("customer_unique_id")["order_date"]
    .diff()
    .dt.days
)

gap_df = (
    avg_gap
    .groupby("customer_unique_id", as_index=False)
    .agg(
        avg_days_between_orders=("gap", "mean")
    )
)
```

```python
gap_df["avg_days_between_orders"] = gap_df["avg_days_between_orders"].fillna(0)
```

```python
customer_features = customer_features.merge(gap_df , on='customer_unique_id',how='left')
```

```python
customer_features = customer_features.drop(
    columns=[
        "first_purchase",
        "last_purchase",
        "customer_active_days"
    ]
)
```

```python
customer_features['churn'] = (customer_features['recency_days']>90).astype(int)
```

# Inventory Agent 
group by product_id
1. historical_sales --> count of product_id
2. avg_daily_sales  --> historical_sales + order_date

```python
master_df.info()
```

```python
inventory_df = order_item_df.merge(
    products_df[['product_id' , 'product_category_name']],
    on = 'product_id',
    how= 'left'
)
```

```python
inventory_df = inventory_df.drop(columns="shipping_limit_date")
```

```python
inventory_df = inventory_df.merge(
    orders_df[['order_id','order_purchase_timestamp','order_status']],
    on = 'order_id',
    how ='left'
)
```

```python
inventory_df = inventory_df[inventory_df['order_status']=='delivered'].reset_index(drop=True)
```

```python
inventory_df[['order_date','order_time']] = inventory_df['order_purchase_timestamp'].str.split(' ',expand=True)
```

```python
inventory_df['order_date'] = pd.to_datetime(inventory_df['order_date'])
```

```python
historical = (inventory_df.groupby('product_id').size().reset_index(name='historical_sales'))
```

```python
inventory_df['order_year'] = inventory_df['order_date'].dt.year
inventory_df['order_month'] = inventory_df['order_date'].dt.month

monthly_sales_df = (inventory_df.groupby(['product_id','order_year','order_month'], as_index=False).agg(
    monthly_sales = ('order_item_id','count')
))
```

```python
daily_sales = (
    inventory_df
    .groupby(
        ["product_id", "order_date"],
        as_index=False
    )
    .agg(
        daily_sales=("order_item_id", "count")
    )
)

avg_daily_sales = (
    daily_sales
    .groupby("product_id", as_index=False)
    .agg(
        avg_daily_sales=("daily_sales", "mean")
    )
)
```

```python
inventory_df = (
    historical
    .merge(avg_daily_sales, on="product_id", how="left")
    .merge(monthly_sales_df, on="product_id", how="left")
)
```

```python
inventory_df
```

```python
# Product with higher historical sales usually has more stock

inventory_df["current_stock"] = (
    inventory_df["historical_sales"] *
    np.random.uniform(0.25, 0.60, len(inventory_df))
).round().astype(int)

inventory_df["current_stock"] = (
    inventory_df["current_stock"]
    .clip(lower=10)
)
```

```python
inventory_df["reorder_level"] = (
    inventory_df["current_stock"] *
    np.random.uniform(0.30, 0.40, len(inventory_df))
).round().astype(int)
```

```python
inventory_df['supplier_id'] = order_item_df['seller_id']
```

```python

seller_ids = inventory_df["supplier_id"].unique()


lead_time_map = {
    seller: np.random.randint(2, 11)   # 2–10 days
    for seller in seller_ids
}

inventory_df["lead_time_days"] = (
    inventory_df["supplier_id"]
    .map(lead_time_map)
)
```

```python
inventory_df["reorder_level"] = (
    inventory_df["current_stock"] *
    np.random.uniform(0.30, 0.40, len(inventory_df))
).round().astype(int)
```

```python
inventory_df["reorder_quantity"] = np.where(
    inventory_df["current_stock"] <= inventory_df["reorder_level"],
    inventory_df["monthly_sales"],0
)
```

```python
warehouse_list = [
    "Mumbai",
    "Delhi",
    "Bangalore",
    "Hyderabad",
    "Pune",
    "Chennai",
    "Ahmedabad",
    "Indore",
    
]

inventory_df["warehouse_location"] = np.random.choice(
    warehouse_list,
    size=len(inventory_df),
    p=[0.22 , 0.13 ,0.10 ,0.10 ,0.15, 0.11 ,0.12, 0.07]
)
```

# Finance Agent


```python
finance_df = master_df[[
    'order_id',
    'order_date',
    'payment_value',
    'total_items',
    'total_freight'
]].copy()
```

```python
finance_df['order_date'] = pd.to_datetime(finance_df['order_date'])
```

```python
finance_df
```

```python
finance_df["order_year"] = finance_df["order_date"].dt.year

finance_df["order_month"] = finance_df["order_date"].dt.month

finance_df["order_quarter"] = finance_df["order_date"].dt.quarter
```

```python
master_df.info()
```

```python
finance_df = (
    finance_df
    .groupby(
        ["order_year", "order_month", "order_quarter"],
        as_index=False
    )
    .agg(
        monthly_revenue=("payment_value", "sum"),
        total_orders=("order_id", "count"),
        total_qty_sold = ('total_items','sum'),
        total_freight_cost = ('total_freight','sum')
    )
)
```

```python
import numpy as np

np.random.seed(42)

finance_df["purchase_cost"] = (
    finance_df["monthly_revenue"] *
    np.random.uniform(0.55, 0.75, len(finance_df))
).round(2)
```

```python
finance_df["marketing_cost"] = (
    finance_df["monthly_revenue"] *
    np.random.uniform(0.03, 0.08, len(finance_df))
).round(2)
```

```python
finance_df["salary_cost"] = np.random.randint(
    50000,
    100001,
    len(finance_df)
)
```

```python
finance_df["rent_cost"] = np.random.randint(
    20000,
    50001,
    len(finance_df)
)
```

```python
finance_df["gross_profit"] = (
    finance_df["monthly_revenue"]
    - finance_df["purchase_cost"]
    -finance_df['total_freight_cost']
)

# Gross Profit = S.P - C.P 
```

```python
finance_df["net_profit"] = (
    finance_df["monthly_revenue"]
    - finance_df["purchase_cost"]
    - finance_df["marketing_cost"]
    - finance_df["salary_cost"]
    - finance_df["rent_cost"]
    -finance_df['total_freight_cost']
)
```

```python
finance_df['profit_margin'] = (finance_df["net_profit"] / finance_df["monthly_revenue"]) *100
```
